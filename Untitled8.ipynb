```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

anfal_df = pd.read_csv('train.csv')
# histogram
plt.figure(figsize=(8, 5))
sns.histplot(anfal_df['SalePrice'], kde=True, color='darkblue')
plt.show()
# heat map
numeric_anfal = anfal_df.select_dtypes(include=[np.number]).columns
plt.figure(figsize=(12, 8))
sns.heatmap(anfal_df[numeric_anfal].corr(), cmap='coolwarm', annot=False)
plt.show()
# scatter plot
plt.figure(figsize=(8, 5))
sns.scatterplot(x=anfal_df['GrLivArea'], y=anfal_df['SalePrice'], color='crimson', alpha=0.6)
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
# quartile 
q1_anfal = anfal_df['SalePrice'].quantile(0.25)
q3_anfal = anfal_df['SalePrice'].quantile(0.75)
iqr_anfal = q3_anfal - q1_anfal
lower_anfal = q1_anfal - 1.5 * iqr_anfal
upper_anfal = q3_anfal + 1.5 * iqr_anfal
# eda 
anfal_df = anfal_df[(anfal_df['SalePrice'] >= lower_anfal) & (anfal_df['SalePrice'] <= upper_anfal)]
# categorize 
anfal_df['PriceCategory_anfal'] = pd.qcut(anfal_df['SalePrice'], q=3, labels=['Low', 'Medium', 'High'])

num_anfal = anfal_df.select_dtypes(include=['int64', 'float64']).columns.drop('SalePrice')
cat_anfal = anfal_df.select_dtypes(include=['object']).columns

for col in num_anfal:
    anfal_df[col] = anfal_df[col].fillna(anfal_df[col].median())

for col in cat_anfal:
    anfal_df[col] = anfal_df[col].fillna(anfal_df[col].mode()[0])

anfal_df = pd.get_dummies(anfal_df, columns=cat_anfal, drop_first=True)

scaler_anfal = StandardScaler()
anfal_df[num_anfal] = scaler_anfal.fit_transform(anfal_df[num_anfal])
# train data 
X_anfal = anfal_df.drop(['SalePrice', 'PriceCategory_anfal'], axis=1)
y_reg_anfal = anfal_df['SalePrice']
y_clf_anfal = anfal_df['PriceCategory_anfal']

X_train_anfal, X_test_anfal, y_train_reg_anfal, y_test_reg_anfal, y_train_clf_anfal, y_test_clf_anfal = train_test_split(
    X_anfal, y_reg_anfal, y_clf_anfal, test_size=0.2, random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import r2_score, accuracy_score
# linear regression
lr_anfal = LinearRegression()
lr_anfal.fit(X_train_anfal, y_train_reg_anfal)
reg_preds_anfal = lr_anfal.predict(X_test_anfal)

r2_anfal = r2_score(y_test_reg_anfal, reg_preds_anfal)

rfc_anfal = RandomForestClassifier(n_estimators=100, random_state=42)
rfc_anfal.fit(X_train_anfal, y_train_clf_anfal)
clf_preds_anfal = rfc_anfal.predict(X_test_anfal)

acc_anfal = accuracy_score(y_test_clf_anfal, clf_preds_anfal)

print("Linear Regression R-squared:", r2_anfal)
print("Random Forest Accuracy:", acc_anfal)
```
